```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
# get the instance of the genrative model
llm  = ChatGoogleGenerativeAI(model= "gemma-4-31b-it")

response = llm.invoke("How many moons does jupiter have?")
print(response.text)
```

```python
# lets give system prompt to the genAI which indicates in what format we want the answer
response = llm.invoke([
    ["system"," You are a helpful assistant that answers in single line"],
    ["human","How many moons does Jupiter have?"],
])

print(response.text)
```

```python
# lets set the temparature value which indicates how creative the llm should be
prompt = "Write a single tagline for quirky coffee shop. Only output the tagline, nothing else"

llm = ChatGoogleGenerativeAI(model = "gemini-3.7-flash",temperature = 0)
response = llm.invoke(prompt)
print(response.text)
```

```python
# lets change the temperature to 1 and see how creatively the model responds
llm = ChatGoogleGenerativeAI(model = "gemini-3.7-flash",temperature = 1)
response = llm.invoke(prompt)
print(response.text)
```

```python
from langchain_groq import ChatGroq
```

```python
# Calling Groq models
llm = ChatGroq(model = "qwen/qwen3.6-27b", reasoning_format="parsed")

messages = [
    (
        'system',
        "You are helpful assistant that gives a short answer to a question",
    ),
    (
        "human",
        "How many moons does Jupiter have?"
    ),
]

ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
