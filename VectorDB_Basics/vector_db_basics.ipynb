```python
import chromadb

client = chromadb.Client()
```

```python
collection = client.create_collection("news_articles")
```

```python
# adding simple single line documents

# by doing this, embeddings will be created for all the documents and is stored in chromadb

collection.add(ids = ["id1","id2","id3","id4"],
documents = [
    "Apple is leading in smart phone game with iphone sales upto by 35%",
    "Tesla booked a minor profit of 1 billion $ in Q2",
    "Apples are high in fiber, Vitamin C and various antioxidants",
    "SpaceX got NASA contract worth 10 billion $",
    ]
)
```

```python
# fect documents and embeddings
data = collection.get(
    include=["documents","embeddings","metadatas"]
)

for i,emb in enumerate(data["embeddings"]):
    print(f"Document {i+1}")
    print("Text:",data["documents"][i])
    print("Embedding length:", len(emb))
    print("First 10 values of embedding:",emb[:10])
```

```python
results = collection.query(
    query_texts = ["this query is related to Elon musk"],
    n_results = 2
)

print(results)
```
