Health Analysis and Diet Recommendation

```python
# Takes in the blood report , analyzes the report and recommend the diet to maintain good health

from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

# initialize the llm model that we will be using to analyze the health report
llm = ChatGoogleGenerativeAI(model = "gemini-3.7-flash")
```

```python
with open("blood_work.txt","r") as f:
    blood_report = f.read()

print(blood_report[:200])
```

```python
# lets use the llm to analyze the health report and give us summary

extraction_prompt = f"""
You are a medical data extraction assistant.

from the blood report below, extract all test values abd classify each one as HIGH, LOW, NORMAL
based on the reference ranges provided in the report.

format your response as:
-Test Name: value | Status: HIGH/LOW/NORMAL | Reference: range

Blood Report:
{blood_report}
"""

extracted_response = llm.invoke(extraction_prompt)
extracted_values = extracted_response.text

print(extracted_values)
```

```python
diet_prompt = f"""
You are a Clinical Dietitian. Based on the blood report extracted values, recommend a indian diet plan to maintain good health.
Provide specific food items and portion sizes for each meal (breakfast, lunch, dinner, and snacks).
Ensure that the diet plan addresses any HIGH or LOW values identified in the blood report.

Blood Report Extracted Values:
{extracted_values}
"""

diet_response = llm.invoke(diet_prompt)
print(diet_response.text)
```
